```python
import numpy as np

# ── columnas que agrega load_qc_tsv, detectadas del df real ──────────────
mean_cols = [c for c in df.columns if c.startswith("ImageQuality_")
             and any(k in c for k in ("PowerLogLogSlope", "MedianIntensity",
                                       "FocusScore"))]
pct_cols  = [c for c in df.columns
             if c.startswith("ImageQuality_PercentMaximal_")
             or c.startswith("ImageQuality_PercentMinimal_")]
sum_cols  = [c for c in df.columns if c.startswith("Count_")]

gkeys = ["Metadata_Plate", "Metadata_Well"]
gkeys = [c for c in gkeys if c in df.columns]

# ── well-level (como load_qc_tsv) ────────────────────────────────────────
well = df.group_by(gkeys).agg(
    [pl.col(c).mean() for c in mean_cols] +
    [pl.col(c).mean() for c in pct_cols] +
    [pl.col(c).sum()  for c in sum_cols]
)

# ── site-level (como load_qc_sites) ──────────────────────────────────────
gkeys_s = gkeys + ["Metadata_Field"]
site = df.group_by(gkeys_s).agg(
    [pl.col(c).mean() for c in mean_cols] +
    [pl.col(c).mean() for c in pct_cols] +
    [pl.col(c).sum()  for c in sum_cols]
)

# ── sanity: ¿cuántos sites por well? ─────────────────────────────────────
per_well = site.group_by(gkeys).len()
print("sites por well — min/max:",
      per_well["len"].min(), "/", per_well["len"].max())

# ── tomar un well de muestra y comparar ──────────────────────────────────
w0 = well.row(0, named=True)
key = {k: w0[k] for k in gkeys}
sub = site.filter(
    pl.all_horizontal([pl.col(k) == v for k, v in key.items()])
)

mcol = mean_cols[0]
ccol = sum_cols[0]

mean_of_sites = sub[mcol].mean()
sum_of_sites  = sub[ccol].sum()

print(f"\nwell de muestra: {key}\n")
print(f"[MEAN] {mcol}")
print(f"   mean(sites) = {mean_of_sites}")
print(f"   well-level  = {w0[mcol]}")
print(f"   MATCH: {abs(mean_of_sites - w0[mcol]) < 1e-6}\n")

print(f"[SUM]  {ccol}")
print(f"   sum(sites) = {sum_of_sites}")
print(f"   well-level = {w0[ccol]}")
print(f"   MATCH: {abs(sum_of_sites - w0[ccol]) < 1e-6}")
```
